An `assert` checks a condition and raises `AssertionError` if it is false. The helper checks each function against an expected answer and checks that it leaves the input unchanged. Agreement between two functions alone would not show that the answer is correct.

`deepcopy` makes an independent copy for these checks. You do not need it in your solutions. Initial test runs catch an unfinished function's `NotImplementedError` and print a reminder. A wrong completed implementation still produces an assertion failure.


````python
import sys
import platform
from copy import deepcopy
from timeit import repeat

print('Python:', sys.version.split()[0])
print('Implementation:', platform.python_implementation())
print('Platform:', platform.platform())
print('Recursion limit:', sys.getrecursionlimit())


def check_pair(recursive, iterative, cases):
    """Check expected answers and unchanged inputs for both functions."""
    for data, expected in cases:
        before = deepcopy(data)
        recursive_result = recursive(data)
        assert data == before, 'Recursive function changed the input'
        assert type(recursive_result) is type(expected), (
            f'Recursive return type is wrong for {data!r}'
        )
        assert recursive_result == expected, (
            f'Recursive result {recursive_result!r}, expected {expected!r}'
        )

        iterative_result = iterative(data)
        assert data == before, 'Iterative function changed the input'
        assert type(iterative_result) is type(expected), (
            f'Iterative return type is wrong for {data!r}'
        )
        assert iterative_result == expected, (
            f'Iterative result {iterative_result!r}, expected {expected!r}'
        )
    print(f'{len(cases)} cases passed for both functions.')
````

The loop keeps a running total. `sum_recursive` waits for a smaller problem's answer before adding. `sum_tail` passes an updated total into the next call. Its `total` parameter is an **accumulator**, meaning the result so far. Call these functions with only the input list in ordinary use.


````python
def sum_iterative(values):
    total = 0
    for value in values:
        total += value
    return total


def sum_recursive(values, index=0):
    if index == len(values):
        return 0
    later_total = sum_recursive(values, index + 1)
    return values[index] + later_total


def sum_tail(values, index=0, total=0):
    if index == len(values):
        return total
    return sum_tail(values, index + 1, total + values[index])


counts = [12, 8, 5]
print('Loop:', sum_iterative(counts))
print('Non-tail recursion:', sum_recursive(counts))
print('Tail recursion:', sum_tail(counts))
````

### D1. Timing the three totals

The helper measures **five batches of 200 calls** with `timeit.repeat`. It reports the shortest batch time divided by 200 in **microseconds per call**. One microsecond is one millionth of a second. The shortest batch reduces the influence of interruptions from other programs.

`lambda: function(data)` is a small function with no arguments that makes the call to measure. You may use it as given. The data is created and its result is checked outside the timer. Printing and tracing are outside the timed call too.

All three totals perform one addition per list item, but the recursive versions also create calls. Record what your machine shows. The tail version is not guaranteed to be faster than the non-tail version.


````python
def microseconds_per_call(function, data, number=200):
    samples = repeat(lambda: function(data), number=number, repeat=5)
    return min(samples) / number * 1_000_000


# Leave room for existing notebook calls. Do not change the recursion limit.
max_recursive_n = min(400, max(0, sys.getrecursionlimit() - 100))
sizes = [n for n in [10, 50, 100, 200, 400] if n <= max_recursive_n]
timing_results = []

for n in sizes:
    values = [1] * n  # Created outside the timed call.
    assert sum_iterative(values) == n
    try:
        assert sum_recursive(values) == n
        assert sum_tail(values) == n
        loop_us = microseconds_per_call(sum_iterative, values)
        non_tail_us = microseconds_per_call(sum_recursive, values)
        tail_us = microseconds_per_call(sum_tail, values)
    except RecursionError:
        print(f'Skipping n={n}: a recursive version reached the current limit.')
        continue
    timing_results.append((n, loop_us, non_tail_us, tail_us))

print(f'{"n":>6} {"loop (us/call)":>16} {"non-tail (us/call)":>20} {"tail (us/call)":>16}')
for n, loop_us, non_tail_us, tail_us in timing_results:
    print(f'{n:>6} {loop_us:>16.3f} {non_tail_us:>20.3f} {tail_us:>16.3f}')

if not timing_results:
    print('No sizes fit this environment. Record this and ask your instructor.')
````

### D2. Observe a depth constraint

Run the next supplied cell. It tries a longer, valid list with all three totals and catches `RecursionError` so the notebook can continue. The input is capped at **2,000 items**. Do not change the recursion limit.

CPython does not eliminate Python tail calls. A tail-recursive function still keeps active calls and can reach the configured depth limit. The distinction between tail and non-tail tells us where the calculation happens, rather than guaranteeing constant stack space in Python.

If a recursive version succeeds because your environment's limit is high, record that observed result and the list length. You do not need to force an error. Use the official [CPython explanation](https://docs.python.org/3/whatsnew/3.14.html#a-new-type-of-interpreter) and [depth-limit documentation](https://docs.python.org/3/library/sys.html#sys.getrecursionlimit) for Python-specific behavior.


````python
long_n = min(sys.getrecursionlimit() + 10, 2000)
long_counts = [1] * long_n
depth_observations = {}

print('Configured limit:', sys.getrecursionlimit())
print('List length:', long_n)
loop_result = sum_iterative(long_counts)
assert loop_result == long_n
depth_observations['loop'] = f'returned {loop_result}'
print('Loop:', depth_observations['loop'])

for label, function in [('non-tail', sum_recursive), ('tail', sum_tail)]:
    try:
        result = function(long_counts)
    except RecursionError:
        depth_observations[label] = 'RecursionError'
    else:
        assert result == long_n
        depth_observations[label] = f'returned {result}'
    print(f'{label}: {depth_observations[label]}')
````
